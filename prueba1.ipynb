```python
# EJERCICIO 2: Cuenta cuántos valores del arreglo son mayores que el promedio.
valores = [10, 55, 23, 8, 90, 33, 47, 12]
promedio = sum(valores)/len(valores)
mayores = 0
for valor in valores:
    if valor > promedio:
        mayores += 1
print("Cantidad de valores mayores que el promedio:", mayores)
```

```python
# EJERCICIO 3: Invierte un arreglo SIN usar reverse() ni slicing [::-1].
def invertir(arreglo):
    for i in range(len(arreglo)//2):
        arreglo[i], arreglo[len(arreglo)-1-i] = arreglo[len(arreglo)-1-i], arreglo[i]
    return arreglo
print(invertir([5, 2, 6, 3, 1]))
```

```python
#implemente una funcion q verifique si un arreglo esta ordenado acendentemente
def orden(arreglo):
    for i in range(len(arreglo)-1):
        if arreglo[i] > arreglo[i+1]:
            return False 
    return True
orden([3, 1, 9, 3, 7])
```
